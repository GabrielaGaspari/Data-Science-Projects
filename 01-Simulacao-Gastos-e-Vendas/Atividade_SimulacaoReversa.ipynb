Vamos primeiro fazer o Gráfico #1 Relação entre Idade e Gasto Mensal (a partir do CVS)

```python
# Primeiro vamos importar as bibliotecas do python
import numpy as np # Utilizaremos o Numpy para gerar os valores
import pandas as pd # O Pandas para gerar e manipular o CVS
import matplotlib.pyplot as plt # E por fim MatplotLib para gerar o gráfico de dispersão

# Abaixo estaremos usando o Numpy para gerar aleatoriamente os valores seguindo alguns padrões, ex: quanto maior a idade maior será o gasto mensal
np.random.seed (42)
idade = np.random.randint (18, 66, 210)
gastos = 2.5 * idade + np.random.normal(0, idade * 0.6, 210)

df = pd.DataFrame({"Idade": idade,"Gasto Mensal (R$)": gastos})

# Aqui estão informções de como o gráfico será feito
plt.plot(df["Idade"], df["Gasto Mensal (R$)"], '.')
plt.xlabel("Idade")
plt.ylabel("Gasto Mensal (R$)")
plt.title("Relação entre Idade e Gasto Mensal (a partir do CVS)")
plt.grid()
plt.show()

# Agora conveter o DF em CVS para poder fazer o download enviar o arquivo
df.to_csv
df.to_csv('Relação entre Idade e Gasto Mensal.csv', float_format='%.2f')
```

Agora vamos para o Gráfico #2

```python
#primeiro vamos importar as bibliotecas do python
import numpy as np # Utilizaremos o Numpy para gerar os valores
import pandas as pd # O Pandas para gerar e manipular o CVS
import matplotlib.pyplot as plt # E por fim MatplotLib para gerar o grafíco
import matplotlib.dates as mdates # e a variação para manipular datas e exibir apenas meses que desejamos

# Abaixo estaremos usando o Numpy para gerar os valores baseado no gráfico enviado
data = pd.to_datetime([ # converte dados de texto em objetos de data e hora
    '2020-01-01', '2020-02-01', '2020-03-01', '2020-04-01', '2020-05-01',
    '2020-06-01', '2020-07-01', '2020-08-01', '2020-09-01', '2020-10-01',
    '2020-11-01', '2020-12-01', '2021-01-01', '2021-02-01', '2021-03-01',
    '2021-04-01', '2021-05-01', '2021-06-01', '2021-07-01', '2021-08-01',
    '2021-09-01', '2021-10-01', '2021-11-01', '2021-12-01', '2022-01-01',
    '2022-02-01', '2022-03-01', '2022-04-01', '2022-05-01', '2022-06-01',
    '2022-07-01', '2022-08-01', '2022-09-01', '2022-10-01', '2022-11-01',
    '2022-12-01'])
vendas = np.array([
    105, 110, 152, 140, 136, 150, 125, 138, 135, 149, 170, 165,
    180, 200, 199, 212, 190, 185, 215, 198, 213, 198, 210, 208,
    225, 255, 256, 278, 270, 272, 265, 285, 265, 258, 285, 276])

df = pd.DataFrame({"Data": data,"Vendas (R$)": vendas})

# Utilizaremos agora a biblioteca dates do matplotlib, Ele examina o intervalo de datas gráfico e escolhe as posições ideais para os marcadores
plt.gca().xaxis.set_major_locator(mdates.AutoDateLocator())
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m'))
plt.gcf().autofmt_xdate()

# Aqui estão informções de como o gráfico será feito

plt.plot(df["Data"], df["Vendas (R$)"])
plt.xlabel("Data")
plt.ylabel("Vendas (R$)")
plt.title("Vendas Mensais com sazonalidade (2020-2022)")
plt.grid()
plt.show()

# Agora conveter o DF em CVS para poder fazer o download enviar o arquivo
df.to_csv('Vendas Mensais com sazonalidade (2020-2022).csv')
```
